```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
import numpy as np

best_mse = float("inf")

for trees in [50, 100, 200]:

    for depth in [5, 10, None]:

        model = RandomForestRegressor(
            n_estimators=trees, max_depth=depth, random_state=42
        )

        scores = cross_val_score(model, X, Y, cv=10, scoring="neg_mean_squared_error")

        mse = (-scores).mean()

        print(f"trees={trees}, depth={depth}, mse={mse:.4f}")

        if mse < best_mse:
            best_mse = mse
            best_trees = trees
            best_depth = depth

print(f"\nBest: trees={best_trees}, depth={best_depth}, mse={best_mse:.4f}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score

model = Pipeline(
    [("scaler", StandardScaler()), ("svr", SVR(kernel="rbf", C=0.55, gamma="scale"))]
)

scores = cross_val_score(model, X, Y, cv=10, scoring="neg_mean_squared_error")

mse = (-scores).mean()

print(mse)
```
